# Isochrone posterior simulation at given ages

`simulate_stars_given_age` は `sim_stars_given_age.py` に関数化されている。入力は真の年齢 `log10(age / yr)` で、各星の isochrone posterior を返す。

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import truncnorm

from sim_stars_given_age import simulate_stars_given_age, stack_isochrone_samples
```

```python
RNG_SEED = 0
N_STARS = 100
LOGAGE_MU = 9.5
LOGAGE_SIGMA = 0.2
LOGAGE_MIN = 8.0
LOGAGE_MAX = np.log10(13.8e9)

lower = (LOGAGE_MIN - LOGAGE_MU) / LOGAGE_SIGMA
upper = (LOGAGE_MAX - LOGAGE_MU) / LOGAGE_SIGMA
logages_true = truncnorm.rvs(
    lower, upper, loc=LOGAGE_MU, scale=LOGAGE_SIGMA,
    size=N_STARS, random_state=RNG_SEED,
)
```

```python
isochrone_posterior = simulate_stars_given_age(
    logages_true,
    rng_seed=RNG_SEED,
    num_warmup=1000,
    num_samples=250,  # 4 chains x 250 = 1000 draws per star
    num_chains=4,
)
age_samples, radius_samples = stack_isochrone_samples(isochrone_posterior)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].hist(10 ** (logages_true - 9.0))
axes[0].set(xlabel="True age [Gyr]", ylabel="Number of stars")
axes[1].hist(age_samples[0], bins=30)
axes[1].axvline(10 ** (logages_true[0] - 9.0), color="black", ls="--", label="truth")
axes[1].set(xlabel="Isochrone age [Gyr]", ylabel="Posterior draws")
axes[1].legend()
fig.tight_layout()
```
